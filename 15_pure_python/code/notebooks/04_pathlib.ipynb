## Why pathlib

Old Python used `os.path` — string-based paths. Modern Python uses `pathlib.Path` — object-based paths.

| Feature | `os.path` | `pathlib` |
|---------|----------|-----------|
| Join paths | `os.path.join('a', 'b')` | `Path('a') / 'b'` |
| Get name | `os.path.basename(p)` | `Path(p).name` |
| Get parent | `os.path.dirname(p)` | `Path(p).parent` |
| Check exists | `os.path.exists(p)` | `Path(p).exists()` |
| Read file | `open(p).read()` | `Path(p).read_text()` |
| Cross-platform | Manual | Automatic |

Find files with patterns — Glob is used to list "all JPGs in a folder".



```python
from pathlib import Path

# Folder with potato images
images_dir = Path(r'C:\Users\USER\rag_course\14_multimodal_rag\data')

# Find all JPG files
jpg_files = list(images_dir.glob('*.JPG'))

print(f'Found {len(jpg_files)} JPG files:')

for f in jpg_files[:5]:
    print(f'{f.name}')
    
print()

# Find all images (jpg, jpeg, png) 
patterns = ['*.jpg', '*.jpeg', '*.png', '*.JPG']
all_images = []

for pattern in patterns:
    all_images.extend(images_dir.glob(pattern))
    print(all_images)

print(f'Found {len(all_images)} total images')
```

Build the ConfigPaths class — one place that knows where every project file lives.

```python
from pathlib import Path

class ConfigPaths:
    '''
    Central place for all AgriVoice file paths.

    Every part of the app asks this class for paths —
    no more hardcoded strings scattered everywhere.
    '''
    
    def __init__(self, project_root: str):
        '''Store the project root and derive every path from it.'''
        
        # Pass only root
        self.root = Path(project_root)
        
    @property
    def chroma_dir(self) -> Path:
        '''Where the Chroma vector store lives.'''
        
        return self.root / 'chroma_db_domain'
    
    @property
    def data_dir(self) -> Path:
        '''Where raw data files (PDFs, HTML, CSV) live.'''
        
        return self.root / '04_data_ingestion_document_processing' / 'data'
    
    @property
    def model_dir(self) -> Path:
        '''Where trained PyTorch models will live.'''
        
        return self.root / 'models'
    
    @property
    def logs_dir(self) -> Path:
        '''Where log files are written.'''
        
        return self.root / 'logs'
    
    @property
    def tests_dir(self) -> Path:
        '''Where test fixtures and sample data live.'''
        
        return self.root / '15_pure_python' / 'agrivoice' / 'tests'
    
    def ensure_dirs(self) -> None:
        '''Create every folder that should exist.'''
        
        for folder in [self.model_dir, self.logs_dir]:
            folder.mkdir(parents=True, exist_ok=True)
        print('All directories ready')
        
    def __Repr__(self) -> str:
        '''Debug representation showing the root.'''
        return f'ConfigPAths(root={self.root})'
    
    
print('ConfigPaths class defined')
```
