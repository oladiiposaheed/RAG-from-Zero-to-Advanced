### Type Hints

Without vs With Type Hints

```python
# Version 1 — no type hints
def add_v1(a, b):
    return a + b


# Version 2 — with type hints
def add_v2(a: int, b: int) -> int:
    return a + b


# Both work the same way
print('v1 result:', add_v1(2, 3))
print('v2 result:', add_v2(2, 3))
```

```python
result = add_v2('hello', 'world')

print('Result:', result)
print('Type:  ', type(result).__name__)
```

Literal for Function Behaviour

```python
from typing import Literal

def format_answer(
    text: str,
    style: Literal['short', 'detailed', 'bullet']
) -> str:
    '''
    Format an answer in one of three fixed styles
    '''
    
    if style == 'short':
        return text[:100]
    elif style == 'detailed':
        return f'Detailed: {text}'
    else:
        return '• ' + '\n• '.join(text.split('. '))
    
answer = 'To treat cassava mosaic, use disease-free cuttings. Plant resistant varieties. Remove infected plants. Finally consult specialist'
print('SHORT:')
print(format_answer(answer, 'short'))
print()
print('BULLET:')
print(format_answer(answer, 'bullet'))
```

Literal for API Request Types

```python
def process_request(
    action: Literal['ask', 'classify', 'translate', 'speak'],
    payload: str,
) -> str:
    '''Route a request to the right handler.'''
    
    handlers = {
        'ask': lambda p: f'RAG: answering "{p}"',
        'classify': lambda p: f'Vision: classifying image "{p}"',
        'translate': lambda p: f'Translate: "{p}" -> Yoruba',
        'speak':     lambda p: f'TTS: speaking "{p}"',
    }
    
    return handlers[action](payload)

print(process_request('ask', 'How do I treat maize smut?'))
print(process_request('classify', 'photo.jpg'))
print(process_request('translate', 'hello'))
```

Callable Types
* Callable — how to hint "this parameter must be a function".
* Callable[[int], int]: a function that takes an int and returns an int

```python
from typing import Callable

def apply_twice(func: Callable[[int], int], value: int) -> int:
    '''Apply a function twice to a value.'''
    
    return func(func(func(value)))
    #return func(value)

# A simple function to test
def add_one(x: int) -> int:
    return x + 1

def double(x: int) -> int:
    return x * 2

print('Add one twice:', apply_twice(add_one, 5))
print('Double twice: ', apply_twice(double, 5))
```

 Square a List, Then Sum and Average

```python
from typing import  Callable

def apply_to_list(
    numbers: list[int],
    transform: Callable[[int], int],
) -> list[int]:
    '''
    Apply a transformation to every number in the list.
    '''
    
    # Loop through each number and apply the transform
    return [transform(n) for n in numbers]

def square(x: int) -> int:
    
    return x * x

def cube(x: int) -> int:
    return x ** 3

def find_sum(numbers: list[int]) -> list:
    return sum(numbers)

def find_average(numbers: list[int]) -> float:
    if not numbers:
        return 0.0
    return sum(numbers) / len(numbers)


# The original list
original = [1, 2, 3, 4, 5]
print(f'Original: {original}')
print()

# Square every number
squared = apply_to_list(original, square)
print(f'Squared:  {squared}')

# Cube every number (proves the transform is swappable)
cubed = apply_to_list(original, cube)
print(f'Cubed:    {cubed}')
print()

# Sum and average of squared
print(f'Sum of squared:     {find_sum(squared)}')
print(f'Average of squared: {find_average(squared)}')
print()

# Sum and average of original
print(f'Sum of original:     {find_sum(original)}')
print(f'Average of original: {find_average(original)}')
```

Protocol

```python
def make_it_speak(animal: Animal) -> str:
    """Refuse to run if the object isn't an Animal."""
    if not isinstance(animal, Animal):
        raise TypeError(f'{type(animal).__name__} is not an Animal')
    return animal.speak()


# Valid call
print(make_it_speak(Dog()))

# Invalid call — caught at runtime
try:
    print(make_it_speak(Rock()))
except TypeError as e:
    print(f'❌ {e}')
```

Protocol — AgriVoice 
* Build a RAGService that uses Protocols for its dependencies

Step 1: The Retriever Protocol

```python
from typing import Protocol, runtime_checkable

# Define the contract for a retriever.
@runtime_checkable
class RetrieverProtocol(Protocol):
    '''
    Retriever is any object with a .fetch(query) method that returns a list of strings
    '''
    
    def fetch(self, query: str) -> list[str]:
        ...
```

Step 2: The LLM Protocol

```python
# Define the contract for an LLM.

@runtime_checkable
class LLMProtocol(Protocol):
    '''
    LLM is any object with a .generate(prompt) method that returns a string
    '''
    
    def generate(self, prompt: str) -> str:
        ...
    
```

Step 3: The RAGService Class

```python
# A service that accepts anything matching the two Protocols.

class RAGService:
    '''
    A RAG service — properly typed with Protocols.
    '''
    
    def __init__(
        self,
        retriever: RetrieverProtocol,
        llm: LLMProtocol,
    ):
        self._retriever = retriever
        self._llm = llm
        self._is_ready = False
        
    def warm_up(self) -> None:
        '''
        Mark the service as ready.
        '''
        self._is_ready = True
        print('RAGService warmed up')
        
    def ask(self, question: str) -> str:
        '''
        Retrieve chunks and generate an answer.
        '''
        if not self._is_ready:
            raise RuntimeError('Call warm_up() first')
        chunks = self._retriever.fetch(question)
        prompt = f'Context: {chunks}. Question: {question}'
        return self._llm.generate(prompt)
        
```

Test the Service

```python
# The Fake Classes
class FakeRetriever:
    '''
    Fake retriever — matches RetrieverProtocol.
    '''
    
    def fetch(self, query: str) -> list[str]:
        return [f'doc about "{query}"']
    
    
class FakeLLM:
    '''
    Fake LLM — matches LLMProtocol.
    '''
    
    def generate(self, prompt: str) -> str:
        return f'Answer: {prompt}'
    
print('Fake classes defined')
```

Wire It Up and Run

```python
# Create the service with the fakes
service = RAGService(
    retriever=FakeRetriever(),
    llm=FakeLLM(),
)

# Warm it up
service.warm_up()

# Ask a question
answer = service.ask('How do I treat cassava mosaic?')
print(f'\n{answer}')
```

Different Classes

```python
class SimpleRetriever:
    '''A different retriever — still matches the Protocol.'''

    def fetch(self, query: str) -> list[str]:
        return ['one doc', 'two docs']


class EchoLLM:
    '''A different LLM — still matches the Protocol.'''

    def generate(self, prompt: str) -> str:
        return f'Echo: {prompt}'


# Same service, different fakes
service2 = RAGService(
    retriever=SimpleRetriever(),
    llm=EchoLLM(),
)
service2.warm_up()
print(service2.ask('Test'))
```
