Method Types

```python
class Dog:
    '''Dog class showing the three method types.'''
    
    species = 'Canis familiaris'
    
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age
        
    # Instance method: uses self, works on instance data
    def bark(self) -> str:
        return f'{self.name} says woof!'
    
    # Class method: uses cls, works on the whole class
    @classmethod
    def from_birth_year(cls, name: str, birth_year: int) -> 'Dog':
        age = 2026 - birth_year
        return cls(name=name, age=age)
    
    # Static method: no self, no cls -> just a utility
    @staticmethod
    def is_puppy(age: int) -> bool:
        '''Check if a dog is a puppy based on age.'''
        return age < 1
```

Test All Three

```python
# Instance method
rex = Dog(name='Rex', age=12)
print('Instance method:  ', rex.bark())

# 2. Class method — called on the class, returns an instance
luna = Dog.from_birth_year(name='Luna', birth_year=2024)
print('Class method:     ', f'Luna is {luna.age} years old')

# 3. Static method — called on the class, no instance needed
print('Static method:    ', f'Is 0 a puppy? {Dog.is_puppy(0)}')
print('Static method:    ', f'Is 3 a puppy? {Dog.is_puppy(3)}')
```

Properties

```python
class Dog:
    '''Dog with a property for age.'''
    
    def __init__(self, name: str, age: int):
        self.name = name
        self._age = age
     
    # getter — runs when read dog.age   
    @property
    def age(self) -> int:
        return self._age
    
    # setter — runs when write dog.age = x
    @age.setter
    def age(self, value: int):
        
        if value < 0:
            raise ValueError('Age cannot be negative')
        self._age = value
        
    @property
    def is_puppy(self) -> bool:
        return self._age < 1
```

Test the Properties

```python
rex = Dog(name='Rex', age=3)

# Read the property
print(f'Age (getter): {rex.age}')
print(f'Is puppy? {rex.is_puppy}')

# Write the property — runs the setter
rex.age = 4
print(f'New age (setter): {rex.age}')

try:
    rex.age = -1
except ValueError as e:
    print(f'Validation error: {e}')
```

**Inheritance**

```python
class Animal:
    '''Base class — shared by all animals.'''
    
    def __init__(self, name: str):
        self.name = name
        
    def speak(self) -> str:
        return f'{self.name} makes a sound'
    

class Dog(Animal):
    '''Dog inherits from Animal — gets name and speak for free'''
    
    def __init__(self, name: str, breed: str):
        
        # call parent's __init__
        super().__init__(name)
        self.breed = breed
        
    def speak(self) -> str:
        
        # override parent behaviour
        return f'{self.name} says woof!'
    

class Cat(Animal):
    '''
    Cat inherits from Animal — overrides speak differently.
    '''
    
    def speak(self) -> str:
        return f'{self.name} says meow!'
    
print('Animal, Dog, and Cat classes defined')
```

Test Inheritance

```python
rex = Dog(name='Rex', breed='Labrador')
whiskers = Cat(name='Whiskers')

print(rex.speak())
print(whiskers.speak())
print()
print(f'Rex name (from Animal):  {rex.name}')
print(f'Rex breed (from Dog):    {rex.breed}')
print(f'Whiskers name:           {whiskers.name}')
print()
print(f'Is Rex a Dog?: {isinstance(rex, Dog)}')
print(f'Is Rex an Animal?: {isinstance(rex, Animal)}')
print(f'Is Whiskers a Dog?: {isinstance(whiskers, Dog)}')
print(f'Is Whiskers a Dog?: {isinstance(whiskers, Animal)}')
```

Calling Parent's Overridden Method

```python
class TalkingDog(Dog):
    '''
    A dog that also greets — demonstrates super() in an overridden method
    '''
    
    def speak(self) -> str:
        parent_speak = super().speak()   # call Dog's speak
        return f'{parent_speak} And also: hello!'
    
talking = TalkingDog(name='Buddy', breed='Beagle')
print(talking.speak())
```

Composition vs Inheritance

```python
# --- COMPOSITION: "has-a" relationship ---
class Engine:
    '''A simple engine.'''
    
    def __init__(self, horsepower: int):
        self.horsepower = horsepower
        
    def start(self) -> str:
        return f'Engine ({self.horsepower}hp) started'
    

class Car:
    '''A Car HAS an Engine — composition.'''
    
    def __init__(self, brand: str, engine: Engine):
        self.brand = brand
        self.engine = engine    # composition
        
    def drive(self) -> str:
        return f'{self.brand}: {self.engine.start()}'
    
    
# INHERITANCE: "is-a" relationship
class Vehicle:
    
    def move(self) -> str:
        return 'moving'
    

class Motorcycle(Vehicle):
    '''A Motorcycle IS A Vehicle — inheritance.'''
    
    def move(self) -> str:
        return 'Motorcycle is moving on two wheels'
    
print('Both patterns defined')
```

Test Both Patterns

```python
# Composition: build an Engine, then pass it to a Car
v9 = Engine(horsepower=500)
my_car = Car(brand='Toyota', engine=v9)

print(my_car.drive())
print(f'My car\'s engine: {my_car.engine.horsepower}hp')
print()

# Inheritance: Motorcycle IS A Vehicle
bike = Motorcycle()
print(bike.move())
print()

# Proof: Car does NOT inherit from Engine
print(f'Is Car an Engine? {isinstance(my_car, Engine)}')
print(f'Is Motorcycle a Vehicle? {isinstance(bike, Vehicle)}')
```

Composition in AgriVoice

```python
class Retriever:
    '''A simple retriever'''
    
    def fetch(self, query: str) -> list:
        '''
        Return a list of document chunks relevant to the query.
        '''
        return [f'chunk about {query}']
    
class LLM:
    '''A simple LLM wrapper.'''
    
    def generate(self, prompt: str) -> str:
        '''Generate a text response from a prompt'''
        return f'Answer to: {prompt}'
    
    
class RAGService:
    '''A RAGService HAS A Retriever and HAS AN LLM — composition.'''
    
    def __init__(self, retriever: Retriever, llm: LLM):
        
        # Store the retriever and LLM so we can call .fetch() and .generate() later
        self.retriever = retriever
        self.llm = llm
        
    def ask(self, question: str) -> str:
        
        # use the retriever to find relevant chunks
        chunks = self.retriever.fetch(question)
        
        # combine chunks + question into a single prompt
        prompt = f'Context: {chunks}. Question: {question}'
        
        #send the prompt to the LLM and return its answer
        return self.llm.generate(prompt)
    
    
service = RAGService(
    retriever=Retriever(),
    llm=LLM(),
)

print(service.ask('How do I treat cassava mosaic?'))
```

**Abstract Base Class — The Contract**

```python
from abc import ABC, abstractmethod

class AgriService(ABC):
    '''
    Abstract base class for every AgriVoice service.
    
    Defines the CONTRACT: every service must implement `run()`.
    
    You cannot create an AgriService() directly — only subclasses.
    '''
    
    def __init__(self, name: str):
        '''# Every service has a name for logging and identification'''

        self.name = name
        
    @abstractmethod
    def run(self, *arg, **kwargs):
        '''Every subclass MUST implement this method.'''
        
        pass
    
    def __repr__(self) -> str:
        # A helpful string representation for debugging
        
        return f'{self.__class__.__name__}(name={self.name!r})'
    
print('AgriService ABC defined')
```

Build real AgriVoice services that follow the contract.

```python
class RetrieverService(AgriService):
    '''
    Retrieves relevant crop and health documents.
    '''
    
    def __init__(self, name: str, vectorstore):
        # Call the parent constructor to set the name
        super().__init__(name)
        
        # Store the vectorstore as a private dependency
        self._vectorstore = vectorstore
        
    def run(self, query: str) -> list:
        '''Fetch chunks for a query — required by AgriService contract.'''
        
        return [f'chunk for: {query}']
    

class VisionService(AgriService):
    '''Classifies crop disease from an image.'''
    
    def __init__(self, name: str, model: str = 'gpt-4o-mini'):
        # Call the parent constructor
        super().__init__(name)
        
        # Store the model name
        self._model = model
        
    def run(self, image_b64: str) -> str:
        '''Classify an image — required by AgriService contract.'''
        
        return 'cassava mosaic disease'
    
    
class TranslationService(AgriService):
    '''Translates text between English and Nigerian languages.'''
    
    def __init__(self, name: str, target_lang: str = 'yoruba'):
        super().__init__(name)
        self._target = target_lang
        
    def run(self, text: str) -> str:
        '''Translate text '''
        
        return f'[translated to {self._target}]: {text}'
    

print('Three services built on the AgriService contract')
```

Test All Services the Same Way

A Document Class With Dunders

```python
class Document:
    '''A document with content and metadata.'''
    
    def __init__(self, content: str, source: str = 'unknown'):
        # The text of the document
        self.content = content
        
        # Where the document came from (file name, URL, etc.)
        self.source = source
        
    def __repr__(self) -> str:
        
        short = self.content[:30] + '...' if len(self.content) > 30 else self.content
        return f'Document(content={short!r}, source={self.source!r})'
    
    def __str__(self) -> str:
        return f'[{self.source}] {self.content}'
    
    def __len__(self) -> int:
        return len(self.content)
    
    def __eq__(self, other) -> bool:
        '''"Two Documents are equal if their content matches.'''
        
        # If the other object isn't a Document, they can't be equal
        if not isinstance(other, Document):
            return False
        
        # Compare the content only (ignore source)
        return self.content == other.content
    
print('Document class defined')
```

Test All Dunders

```python
# Create two Documents
doc1 = Document(
    content='Cassava Mosaic Disease is a viral disease affecting cassava plants.',
    source='crop_disease.pdf',
)
doc2 = Document(
    content='Cassava Mosaic Disease is a viral disease affecting cassava plants.',   # same content
    source='agriculture.html',                                                       # different source
)
doc3 = Document(
    content='Rice Blast is a fungal disease.',                                       # different content
    source='crop_disease.pdf',
)

print('--- repr(doc1) — developer view ---')
print(repr(doc1))
print()

print('--- print(doc1) — user view (uses __str__) ---')
print(doc1)
print()

print('--- len(doc1) — uses __len__ ---')
print(f'Length: {len(doc1)} chars')
print()

print('--- doc1 == doc2 (same content, different source) ---')
print(f'Equal? {doc1 == doc2}')
print()

print('--- doc1 == doc3 (different content) ---')
print(f'Equal? {doc1 == doc3}')
print()

print('--- doc1 == "not a Document" ---')
print(f'Equal? {doc1 == "some string"}')
```

__iter__ — Making Documents Iterable

```python
class Document:
    '''Document with iteration support added'''
    
    def __init__(self, content: str, source: str = 'unknown'):
        
        # The text of the document
        self.content = content
        # Where the document came from
        self.source = source
        
    def __iter__(self):
        '''Iterate over words in the content.'''
        
        # Split content on whitespace and return an iterator
        return iter(self.content.split())
    
    def __len__(self) -> int:
        '''Number of words in the document.'''
        
        # Count words, not characters
        return len(self.content.split())
    
    
doc = Document(
    content='Cassava Mosaic Disease is a viral disease.',
    source='test.pdf'
)

print(f'Words: {len(doc)}')
print()
print('Iterating over words:')
for word in doc:
    print(f'  {word}')
```

### Mini-Project — The RAGService Class

```python
class RAGError(Exception):
    '''Custom exception for RAG errors.'''
    
    pass


class RAGService:
    '''
    Production RAG service using composition.

    HAS A retriever and HAS AN LLM.
    '''
    
    def __init__(self, retriever, llm):
        '''Store the dependencies.'''
        
        # Private attributes — accessed via properties
        self._retriever = retriever
        self._llm = llm
        self._request_count = 0
        self._is_ready = False
        
    @property
    def is_ready(self) -> bool:
        '''True once warm_up() is called'''
        
        return self._is_ready
    
    @property
    def request_count(self) -> int:
        '''How many questions answered.'''
        
        return self._request_count
    
    def warm_up(self) -> None:
        '''Mark the service as ready.'''
        
        self._is_ready = True
        print('RAGService warmed up')
        
    def ask(self, question: str) -> str:
        '''Answer a question using the retriever and LLM.'''
        
        # Refuse if not warmed up
        if not self._is_ready:
            raise RAGError('Call warm_up() first')
        
        # Step 1 — retrieve chunks
        chunks = self._retriever.fetch(question)
        
        # Step 2 — build prompt
        prompt = f'Context: {chunks}. Question: {question}'
        
        # Step 3 — generate answer
        answer = self._llm.generate(prompt)
        
        # Step 4 — track request
        self._request_count += 1
        
        # Step 5 — return
        return answer
    
print('RAGService class defined')
```
