```python
from langchain_openai import ChatOpenAI, OpenAI
from dotenv import load_dotenv
```

```python
# Load API keys from .env
load_dotenv()
```

```python
# Create a Chat Model instance
chat_model = ChatOpenAI(model='gpt-4o-mini', temperature=0)

# Create an LLM (completion) 
llm = OpenAI(model='gpt-3.5-turbo-instruct', temperature=0)

print('Both model instances created.')
print(f'Chat Model type: {type(chat_model).__name__}')
print(f'LLM type: {type(llm).__name__}')
```

Step 2: Invoke the LLM (Text Completion)

```python
# Invoke the LLM with a plain string prompt
llm_response = llm.invoke('What is the cpital of Osun state and Nigeria?')

print('LLM Response')
print(llm_response)
print(type(llm_response))
```

```python
# Invoke the LLM with a plain string prompt
llm_response = llm.invoke('What is the cpital of France?')

print('LLM Response')
print(llm_response)
print(type(llm_response))
```

Step 3: Invoke the Chat Model (Message-Based)

```python
# Chat models require a list of messages
from langchain_core.messages import HumanMessage, SystemMessage

# Create a list of messages
messages = [
    SystemMessage(content='You are a helpful assistant'),
    HumanMessage(content='What is the capital of Osun and Nigeria?')
]

# Invoke the chat model with the list of messages
chat_response = chat_model.invoke(messages)

print('Chat Model Response:')
print(chat_response)
print('\nType:', type(chat_response))
print(f'Content: {chat_response.content}')
```

## LLM vs Chat Model – Key Differences

| Feature | LLM (OpenAI) | Chat Model (ChatOpenAI) |
|---------|--------------|--------------------------|
| **Input** | Single string | List of messages (system, human, assistant) |
| **Output** | String | AIMessage object (contains content + metadata) |
| **Use case** | Text completion, simple prompts | Conversational AI, RAG, structured interactions |
| **LangChain class** | `OpenAI` | `ChatOpenAI` |
| **Example model** | `gpt-3.5-turbo-instruct` | `gpt-4o-mini` |

**Why we use Chat Models in RAG:** They allow us to include system instructions, retrieved context, and user question in separate messages, which improves control and accuracy.
