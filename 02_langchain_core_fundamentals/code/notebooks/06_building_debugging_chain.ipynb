```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
```

```python
load_dotenv()
```

Step 2: Build and Run the Complete Chain

```python
# Create the chat model
llm = ChatOpenAI(model='gpt-4o-mini', temperature=0)

# Create a prompt template with two placeholders
prompt = ChatPromptTemplate.from_messages([
    ('system', 'You are a helpful assistant.'),
    ('human', 'Explain {topic} to a {audience}')
])

# Build the chain: prompt -> model -> parser
chain = prompt | llm | StrOutputParser()

# Invoke the chainwith both inputs
res = chain.invoke({'topic': 'RAG', 'audience': 'beginner'})

print('Result:')
print(res)
print(f'Type: {type(res)}')
```

Step 3: Add Metadata and Tags

**Metadata** is a dictionary of custom key-value pairs (e.g., user ID, session ID, version). Tags are simple strings used for filtering.

```python
# Define custom metadata and tags

metadata = {
    'user_id': 'student_123',
    'session_id': 'module2',
    'experiment': 'basic_chain'
}

tags = ['inro', 'audience', 'opeai']

# Invoke the chain with config
res = chain.invoke(
    {'topic': 'RAG', 'audience': 'beginner'},
    config={'metadata': metadata, 'tags': tags}
)

print('Result')
print(res)
```
