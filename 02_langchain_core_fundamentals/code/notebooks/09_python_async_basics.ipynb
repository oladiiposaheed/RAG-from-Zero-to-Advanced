Step 1: A Simple Async Function

```python
import asyncio

async def say_hello():
    print('Hello')
    await asyncio.sleep(1)
    print('World')
    await asyncio.sleep(1)
    print('Welcome to our Python Class')
    
await say_hello()
```

Step 2: Synchronous vs Asynchronous Timing

```python
# Synchronous version (using time.sleep)

import time
def sync_task(name):
    print(f'{name} start')
    time.sleep(2)   # blocks the whole program for 2 seconds
    print(f'{name} end')
    
# Run two tasks sequentially
start = time.time()
sync_task('Python')
sync_task('FastAPI')
end = time.time()

print(f'Total sync time: {end - start:.2f} seconds')
```

```python
# Asynchronous version (using asyncio.sleep)
import asyncio
import time

async def async_task(name):
    print(f'{name} start')
    await asyncio.sleep(2) # pause, but allow other tasks to run
    print(f'{name} end')
    
async def main():
    # Run two tasks concurrently
    await asyncio.gather(async_task('Python'), async_task('FastAPI'))
    
start = time.time()
await main()
end = time.time()
print(f'Total async time: {end - start:.2f} seconds')
```

Step 3: Creating Tasks

```python
# Create tasks to run coroutines concurrently:
async def main_tasks():
    task_1 = asyncio.create_task(async_task('Python'))
    task_2 = asyncio.create_task(async_task('FastAPI'))
    await task_1
    await task_2

await main_tasks()
```

```python
# Step 4: Async with LangChain
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# Create a chat model
llm = ChatOpenAI(model='gpt-4o-mini', temperature=0)

# Create a prompt template
prompt = ChatPromptTemplate.from_messages([
    ('system', 'You are a helpful assistant.'),
    ('human', 'Explain {topic} in one sentence.')
])

# Build a synchronous-capable chain, but use async methods
chain = prompt | llm | StrOutputParser()

# async wrapper function to get an answer for a topic
async def get_answer(topic):
    return await chain.ainvoke({'topic': topic})

# main coroutine that runs multiple get_answer concurrently
async def main_answers():
    topics = ['Python', 'FastAPI', 'Django', 'React', 'TypeScript', 'NextJS']
    tasks = [get_answer(t) for t in topics]
    results = await asyncio.gather(*tasks)
    
    for t, res in zip(topics, results):
        print(f'{t}: {res}')
  
# Run the main_answers coroutine      
await main_answers()
```
