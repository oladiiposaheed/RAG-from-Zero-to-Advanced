```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
```

```python
load_dotenv()
```

Step 2: Streaming Tokens

```python
# Create model and prompt
llm = ChatOpenAI(model='gpt-4o-mini', temperature=0)

prompt = ChatPromptTemplate.from_messages([
    ('system', 'You are a helpful assistant.'),
    ('human', 'Explain {topic} in two sentences.')
])

# Build chain with parser
chain = prompt | llm | StrOutputParser()

# Use stream instead of invoke
print('Streaming response:')

for chunk in chain.stream({'topic': 'RAG'}):
    print(chunk, end='', flush=True)
print()
```

```python
import asyncio

# Define an async function that calls ainvoke and astream
async def run_async_chain():
    
    # Create model and prompt
    llm = ChatOpenAI(model='gpt-4o-mini', temperature=0)
    
    prompt = ChatPromptTemplate.from_messages([
        ('system', 'You are a helpful assistant.'),
        ('human', 'Briefly explain {topic} in one sentence.')
    ])
    
    chain = prompt | llm | StrOutputParser()
    
    # 1. Async invoke
    res = await chain.ainvoke({'topic': 'RAG'})
    print('Async invoke result:')
    print(res)
    print('*' * 100)
    
    # 2. Async stream
    print('Async stream:')
    async for chunk in chain.astream({'topic': 'RAG'}):
        print(chunk, end='', flush=True)
    print('\nDone')
    
# Run the async function
await run_async_chain()
```
