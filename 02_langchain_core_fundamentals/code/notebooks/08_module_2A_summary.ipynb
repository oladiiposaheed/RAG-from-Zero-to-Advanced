# 🧠 Module 2 Part A – LangChain Core Fundamentals Summary


## 1. LLMs vs Chat Models

| Feature | LLM (Text Completion) | Chat Model |
|---------|----------------------|------------|
| Input | Single string | List of messages (system, human, assistant) |
| Output | String | AIMessage object with content + metadata |
| Example | `OpenAI` | `ChatOpenAI` |
| Use case | Simple completion | Conversational AI, RAG |

✅ **We use Chat Models in RAG** because they handle structured prompts and context more naturally.

---

## 2. ChatOpenAI Parameters

- `temperature` → randomness (0 = deterministic, 1 = creative)
- `max_tokens` → maximum output length
- `top_p` → nucleus sampling (alternative to temperature)
- `timeout` → max wait time for API response
- `max_retries` → number of automatic retries on failure

📌 **For RAG**, use low temperature (0–0.3) and reasonable `max_tokens`.

---

## 3. Prompt Templates

- Basic placeholders: `{topic}`
- Multiple placeholders: `{topic}`, `{audience}`, `{word_count}`
- Partial variables: pre‑fill values using `.partial()`
- `MessagesPlaceholder`: insert a list of messages (e.g., chat history)
- `format_messages()` → only prepares messages  
- `.invoke()` → actually calls the model and returns the answer

---

## 4. Output Parsers (Basic)

| Parser | Output | Use Case |
|--------|--------|----------|
| `StrOutputParser` | Plain string | Simple text answers |
| `CommaSeparatedListOutputParser` | Python list | Tags, categories, entity lists |

---

## 5. LCEL Basics

- **Pipe (`|`)** → connects components in sequence
- **`RunnablePassthrough`** → passes input unchanged  
  - Use it to keep original input or extract a field with `lambda x: x['key']`
- **`RunnableLambda`** → wrap custom Python functions
- **`RunnableParallel`** → run multiple branches at once

---

## 6. Building and Debugging a Chain

- Build a complete chain: `prompt | llm | parser`
- Add metadata and tags to runs
- Use LangSmith to inspect traces and debug

---

## 7. Streaming and Async

- `.stream()` → receive output token‑by‑token (real‑time)
- `ainvoke()` → async version of `invoke()`
- `astream()` → async version of `stream()`
