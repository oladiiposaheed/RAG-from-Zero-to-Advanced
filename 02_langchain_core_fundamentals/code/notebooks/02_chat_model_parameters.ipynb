### Task 1.2: Using ChatOpenAI with Parameters


```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage, SystemMessage
from dotenv import load_dotenv
```

```python
# Load environment variables
load_dotenv()
```

Step 2: Create Models with Different Temperatures

```python
# Model A: deterministic (temperature=0)
model_deterministic = ChatOpenAI(model='gpt-4o-mini', temperature=0)

# Model B: creative (temperature=1)
model_creative = ChatOpenAI(model='gpt-4o-mini', temperature=1)

print(f'Model A (temperature={model_deterministic.temperature})')
print(f'Model B (temperature={model_creative.temperature})')
```

Step 3: Compare Responses with Different Temperatures

```python
# Define messages
messages = [
    SystemMessage(content='You are a helpful assistant'),
    HumanMessage(content='Describe a suuny day in one sentence.')
]

# Invoke deterministic model
response_deterministic = model_deterministic.invoke(messages)

# Invoke creative model
response_creative = model_creative.invoke(messages)

# Get response
print('Deterministic (temperature=0):')
print(response_deterministic.content)
print('\nCreative (temperature=1):')
print(response_creative.content)
```

```python
# Define messages
messages = [
    SystemMessage(content='You are a helpful assistant'),
    HumanMessage(content='List the elements in group 7')
]

# Invoke deterministic model
response_deterministic = model_deterministic.invoke(messages)

# Invoke creative model
response_creative = model_creative.invoke(messages)

# Get response
print('Deterministic (temperature=0):')
print(response_deterministic.content)
print('\nCreative (temperature=1):')
print(response_creative.content)
```

Step 4: Use max_tokens to Limit Response Length

```python
# Model C: limited to 20 tokens
model_short = ChatOpenAI(model='gpt-4o-mini', temperature=0, max_tokens=50)

# Model D: allowed up to 500 tokens
model_long = ChatOpenAI(model='gpt-4o-mini', temperature=0, max_tokens=500)

print(f'Model C max_tokens: {model_short.max_tokens}')
print(f'Model D max_tokens: {model_long.max_tokens}')
```

Step 5: Compare Responses with Different max_tokens

```python
# Define messages
messages = [
    SystemMessage(content='You are a helpful assistant.'),
    HumanMessage(content='Explain the important of water to human life.')
]

# Invoke model with max_tokens=50 (short)
response_short = model_short.invoke(messages)

# Invoke model with max_tokens=500 (long)
response_long = model_long.invoke(messages)

# Store responses with their labels
responses = [
    ('Short (max_tokens=50)', response_short),
    ('Long (max_tokens=500)', response_long)
]

for label, response in responses:
    print(f'\n{label}:')
    print('*' * 50)
    print(response.content)
    print('*' * 50)
    
```

Step 6: Create Models with Different top_p Values

```python
# Create Model E with focused (top_p=0.1)
model_focused = ChatOpenAI(model='gpt-4o-mini', top_p=0.1)

# Model F: diverse (top_p=0.9)
model_diverse = ChatOpenAI(model='gpt-4o-mini', top_p=0.9)

print(f'Model E top_p: {model_focused.top_p}')
print(f'Model F top_p: {model_diverse.top_p}')
```

Step 7: Compare Responses with Different top_p Values

```python
# Define messages
messages = [
    SystemMessage(content='You are a helpful assistant.'),
    HumanMessage(content='Who is the present Nigeria Presidient?')
]

# Invoke focused model
response_focused = model_focused.invoke(messages)

# Invoke diverse model
response_diverse = model_diverse.invoke(messages)

# Get both responses with labels
responses = [
    ('Focused', response_focused),
    ('Diverse', response_diverse)
]

for label, response in responses:
    print(f'\n{label}')
    print('*' * 50)
    print(response.content)
    print('*' * 50)
```

Step 8: Create a Model with timeout and max_retries

Parameters: timeout and max_retries

* timeout – how many seconds to wait for a response before raising an error.
* max_retries – how many times to automatically retry the request if it fails (e.g., due to rate limit or temporary network issue).

```python
# Model with timeout and retry settings
model_reliable = ChatOpenAI(
    model='gpt-4o-mini',
    temperature=0,
    timeout=30,    # wait up to 30 seconds
    max_retries=2  # retry up to 2 times on failure
)

print(f'Timeout: {model_reliable.request_timeout} seconds')
print(f'Max retries: {model_reliable.max_retries}')
```

## Summary of ChatOpenAI Parameters

- **temperature**: controls randomness (0 = deterministic, 1 = creative).
- **max_tokens**: limits the number of tokens generated.
- **top_p**: restricts sampling to a cumulative probability mass.
- **timeout**: maximum wait time for API response.
- **max_retries**: number of automatic retries on failure.

For RAG, we typically use low temperature (0–0.3), a reasonable `max_tokens` limit, and reliability settings like `timeout` and `max_retries`.
