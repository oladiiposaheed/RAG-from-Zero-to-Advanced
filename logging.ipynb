### Python Logging – Key Concepts Explained

In this notebook, you will learn:
- What the root logger is
- How to set logging levels
- How handlers work
- Why we clear handlers
- How child loggers propagate to root

```python
import logging

# Get the root logger (no name passed)

root = logging.getLogger()

print(f'Root logger: {root}')
print(f'Default levels: {root.level}')
print(f'Number of handlers: {len(root.handlers)}')
```

### Setting the Level

Levels (low to high):
- DEBUG = 10
- INFO = 20
- WARNING = 30
- ERROR = 40
- CRITICAL = 50

```python
root.setLevel('DEBUG')
print(f'Root level after setLevel("DEBUG"): {root.level}')
```

### Handlers

Handlers send logs to a destination (console, file, etc.).
The root logger needs at least one handler to output logs.

```python
# Add a console handler
console_handler = logging.StreamHandler()
root.addHandler(console_handler)

print(root)
print(f'Number of handlers after adding console: {len(root.handlers)}')
```

### Duplicate Handlers

If you call setup_logging() again without clearing, you'd add another handler,
causing duplicate output. That's why we clear first.

```python
# Clear handlers
if root.hasHandlers():
    root.handlers.clear()
    
print(f'Number of handlers after clear: {len(root.handlers)}')
```

### Propagation from Child Loggers

When you create a logger with a name (e.g., `__name__`), it becomes a child of root.
If the child has no handlers, it propagates its messages to the root logger.

Create child logger and log

```python
# Add a fresh console handler to root

root.addHandler(logging.StreamHandler())

# Create a child logger
child_logger = logging.getLogger('app.services.document_loader')
child_logger.setLevel('DEBUG')

# Log a message from the child
child_logger.info('This is from child logger')
child_logger.debug('Debugging from child')
```
