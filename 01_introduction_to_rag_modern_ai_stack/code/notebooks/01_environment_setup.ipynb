# Module 1 – Environment Setup

In this notebook, we will:

1. Load environment variables from the `.env` file.
2. Verify the Python version.
3. Import the required libraries and check their versions.

> **Prerequisite:** Make sure your virtual environment `rag_env` is activated and you have run `pip install -r requirements.txt` from the course root.

## 1. Load Environment Variables

The `.env` file (located at the course root) stores your API keys and configuration.  
We use `dotenv` to load them into the environment.

```python
import os
from dotenv import load_dotenv
```

```python
# Load .env
load_dotenv()

# Check for required keys
required_keys = ['OPENAI_API_KEY', 'LANGSMITH_API_KEY', 'LANGCHAIN_TRACING_V2', 'LANGCHAIN_PROJECT']

for key in required_keys:
    value = os.getenv(key)
    
    if value:
        masked = value[:4] + '....' + value[-4:] if len(value) > 8 else '****'
        print(f'{key}: {masked}')
    else:
        print(f'{key}: NOT SET')
```

## 2. Check Python Version

```python
import sys

print(f'Python version: {sys.version}')
print(f'Version info: {sys.version_info}')

major, minor = sys.version_info[:2]
if (major, minor) < (3, 9):
    print('Python version is too old. Need >= 3.9')
else:
    print(f'{major}.{minor}')
    print('Python version is sufficient.')
```

```python
import importlib

# List of libraries required for Module 1

libs = [
    'langchain', 'langchain_openai', 'langgraph', 'langsmith', 'dotenv', 'pydantic'
    ]

for lib in libs:
    try:
        mod = importlib.import_module(lib)
        version = getattr(mod, '__version__', 'unknown')
        print(f'{lib} ({version})')
    
    except ImportError as e:
        print(f'{lib}: {e}')
```
