Step 1: Imports and Confirm Tracing Configuration

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from dotenv import load_dotenv
import os
```

```python
# Load environment variables from .env
load_dotenv()

# Check for tracing 
tracing_enabled = os.getenv('LANGCHAIN_TRACING_V2')
print(f'LANGCHAIN_TRACING_V2: {tracing_enabled}')

# Check if LangSmith API key is set
langsmith_key = os.getenv('LANGSMITH_API_KEY')

if langsmith_key:
    masked = langsmith_key[:4] + '....' + langsmith_key[-4:]
    print(f'"LANGSMITH_API_KEY: {masked}')
    
else:
    print('LANGSMITH_API_KEY: NOT SET')
    
```

Step 2: Create and Run a Simple Chain (Traced)

```python
# Create a chat model
llm = ChatOpenAI(model='gpt-4o-mini', temperature=0)

# Define a prompt template with a topic placeholder
prompt = ChatPromptTemplate.from_messages([
    ('system', 'You are a helpful assistant.'),
    ('human', 'Explain {topic} in one sentence.')
])

# Build the chain using LCEL
chain = prompt | llm

# Invoke the chain
response = chain.invoke({'topic': 'LangSmith'})

print('LLM RESPONSE')
print(response.content)
```

Step 3: Add Custom Metadata and Tags

```python
# Define custom metadata and tags
custom_metadata = {
    'user_id': 'studnt_id',
    'session_id': 'module1',
    'experiment': 'tracking_demo'
}

custom_tags = ['intro', 'langsmith', 'openai']

# Invoke the chain with config to attach metadata and tags
response_with_config = chain.invoke(
    {'topic': 'RAG'},
    config={'metadata': custom_metadata, 'tags': custom_tags}
)

print('LLM Response (with metadata/tags):')
print(response_with_config.content)
```
