```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from dotenv import load_dotenv
```

```python
# Load environment variables (API keys)
load_dotenv()

# Create a ChatOpenAI model with gpt-4o-mini
llm = ChatOpenAI(model='gpt-4o-mini', temperature=0)

print(f'Chat model created: {llm.model_name}')
print(f'Temperature: {llm.temperature}')
```

Step 3: Define a Prompt Template

A prompt template is a reusable structure for the messages we send to the LLM. It contains placeholders that we can fill with different values each time.

We'll create a simple prompt with:

* A system message that sets the assistant's behavior.

* A human message that contains the user's question, including a {topic} placeholder.

```python
# Create a chat prompt template with a system and human message
# The human message has a placeholder {topic} that we can fill later

prompt = ChatPromptTemplate.from_messages([
    ('system', 'You are a helpful assistant'),
    ('human', 'Explain {topic} in one sentence.')
])

# Print the structure of the prompt to confirm it was created
print('Prompt template created.')
print(prompt.messages)
print()
print(f'System message: {prompt.messages[0].prompt.template}')
print()
print(f'Human messages: {prompt.messages[1].prompt.template}')
```

Step 4: Build a Chain Using LCEL
Now combine the prompt and the LLM into a single chain using the pipe operator (|). This is LangChain Expression Language (LCEL).

The chain will:

* Take our prompt template.

* Pass its output to the LLM.

* Return the LLM's response

The pipe (|) means: take prompt output and feed it to llm

```python
# Combine the prompt and the LLM into a chain using LCEL

chain = prompt | llm
print('Chain created successfully.')
```

Step 5: Invoke the Chain with a Real Input

Now run the chain with a value for {topic}. The chain will format the prompt, call the LLM, and return a response.

* Invoke the chain with a dictionary containing the topic value
* The chain will:
  1. Fill {topic} with "RAG"
  2. Send the formatted messages to the LLM
  3. Return the generated answer

```python
response = chain.invoke({'topic': 'RAG'})
print('LLM Response:')
print(response.content)
```

Step 6: View the Formatted Prompt Messages

We can use prompt.format_messages() to see the exact messages that get sent to the LLM. 

```python
# Format the prompt with a specific topic value
formatted_messages = prompt.format_messages(topic='RAG')

print("Formatted messages sent to LLM:")
for message in formatted_messages:
    print(f'[{message.type}] {message.content}')
```

Step 7: Invoke the Chain with a Different Topic

Call the chain again with a different value for {topic}

```python
# Invoke the chain with a different topic
response_2 = chain.invoke({'topic': 'Who won 2026 FIFA world cup and which country host it?'})

print("LLM Response for 'FiFA World Cup 2026':")
print(response_2.content)
```
