<p style="margin: 10px 0;">Every pipeline so far has a <strong>fixed sequence</strong>: step 1, step 2, step 3. <strong>You</strong> decide the flow. The LLM just fills in one step at a time.</p>

<h3 style="font-size: 19px; margin: 22px 0 10px 0; color: #2c5282;">What an Agent Is</h3>

<p style="margin: 10px 0;">An <strong>agent</strong> is an LLM that <strong>decides its own flow</strong>. You don't tell it step 1, step 2, step 3. You give it:</p>

<ol style="margin: 10px 0; padding-left: 26px;">
  <li><strong>A goal</strong> — <em>"Help the user diagnose their crop."</em></li>
  <li><strong>Tools</strong> — a set of functions it can call.</li>
  <li><strong>Freedom</strong> — it decides which tools to use, in what order, how many times.</li>
</ol>

<p style="margin: 10px 0;">The agent <strong>plans the path itself</strong>.</p>

<h3 style="font-size: 19px; margin: 22px 0 10px 0; color: #2c5282;">A Concrete Example</h3>

<p style="margin: 10px 0;"><strong>Without an agent (your current RAG):</strong></p>

<pre style="font-size: 14px; line-height: 1.5; margin: 10px 0; padding: 14px 18px; background: #2d3748; color: #f7fafc; border-radius: 6px; overflow-x: auto;"><code>User: "My cassava has yellow leaves. What should I do?"
       │
       ▼
  Always: retrieve → answer</code></pre>

<p style="margin: 10px 0;"><strong>With an agent:</strong></p>

<pre style="font-size: 14px; line-height: 1.5; margin: 10px 0; padding: 14px 18px; background: #2d3748; color: #f7fafc; border-radius: 6px; overflow-x: auto;"><code>User: "My cassava has yellow leaves. What should I do?"
       │
       ▼
  Agent thinks: "I don't have a photo. Should I ask for one?
                 Or search the knowledge base first?"
       │
       ▼
  Agent decides: "Search knowledge base for yellow cassava leaves"
       │
       ▼
  Agent gets results → thinks: "There are multiple possibilities.
                                Should I ask clarifying questions?"
       │
       ▼
  Final answer</code></pre>

<p style="margin: 10px 0;">The agent <strong>chooses the path</strong> based on what it finds.</p>

<h3 style="font-size: 19px; margin: 22px 0 10px 0; color: #2c5282;">What is a "Tool"?</h3>

<p style="margin: 10px 0;">A <strong>tool</strong> is just a Python function that the LLM can ask to run.</p>

<pre style="font-size: 14px; line-height: 1.5; margin: 10px 0; padding: 14px 18px; background: #2d3748; color: #f7fafc; border-radius: 6px; overflow-x: auto;"><code>def get_weather(city: str) -&gt; str:
    """Return the current weather for a city."""
    return "Sunny, 28°C"</code></pre>

<p style="margin: 10px 0;">You describe the function to the LLM:</p>

<blockquote style="margin: 10px 0 10px 18px; padding: 10px 14px; border-left: 4px solid #3182ce; background: #ebf8ff; color: #1a365d;"><em>"Here's a function called <code>get_weather</code> that takes a city and returns the weather."</em></blockquote>

<p style="margin: 10px 0;">The LLM can then decide: <em>"The user asked about tomorrow's weather — I should call <code>get_weather('Lagos')</code>."</em></p>

<p style="margin: 10px 0;">The LLM doesn't run the function itself. It says <strong>"call this function with these arguments"</strong>. Your code actually runs it.</p>

<h3 style="font-size: 19px; margin: 22px 0 10px 0; color: #2c5282;">The Agent Loop (ReAct)</h3>

<p style="margin: 10px 0;">The most common agent pattern is <strong>ReAct</strong> — short for <strong>Reasoning + Acting</strong>:</p>

<pre style="font-size: 14px; line-height: 1.5; margin: 10px 0; padding: 14px 18px; background: #2d3748; color: #f7fafc; border-radius: 6px; overflow-x: auto;"><code>1. THINK:   "What should I do next?"
2. ACT:     Call a tool (or answer directly)
3. OBSERVE: Look at the result
4. Repeat until done</code></pre>

<p style="margin: 10px 0;"><strong>Example with our cassava farmer:</strong></p>

<pre style="font-size: 14px; line-height: 1.5; margin: 10px 0; padding: 14px 18px; background: #2d3748; color: #f7fafc; border-radius: 6px; overflow-x: auto;"><code>THINK: The user says yellow leaves. Let me search the knowledge base.
ACT:   rag_lookup("cassava yellow leaves")
OBSERVE: "Possibilities: Cassava Mosaic Disease, nutrient deficiency"

THINK: I need more info. Let me ask the user for a photo.
ACT:   ask_user("Can you upload a photo?")
OBSERVE: [User uploads photo]

THINK: Let me analyze this photo.
ACT:   vision_model(photo)
OBSERVE: "Cassava Mosaic Disease (94% confidence)"

THINK: Now I have enough to answer.
ANSWER: "Your cassava has Cassava Mosaic Disease. Here's how to treat it..."</code></pre>

<p style="margin: 10px 0;"><strong>That's an agent</strong> — it decides the flow at each step.</p>
