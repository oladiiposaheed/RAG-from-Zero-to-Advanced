<h2 style="font-size: 24px; margin: 4px 0 16px 0; color: #1a365d; border-bottom: 2px solid #3182ce; padding-bottom: 8px;">Module 11 · Notebook 03 — The ReAct Pattern</h2>

<h3 style="font-size: 19px; margin: 22px 0 10px 0; color: #2c5282;">Concept: What Is ReAct?</h3>

<p style="margin: 10px 0; color: #1a365d;">ReAct = <strong>Reasoning + Acting</strong>.</p>

<p style="margin: 10px 0; color: #1a365d;">It's a way of making the LLM <strong>think out loud</strong> in a structured text format, so you can read its reasoning step by step.</p>

<h3 style="font-size: 19px; margin: 22px 0 10px 0; color: #2c5282;">The Two Agent Styles We've Seen</h3>

<h4 style="font-size: 16px; margin: 14px 0 6px 0; color: #2c5282;">Style 1 — Tool-calling agent (Notebooks 01–02)</h4>

<p style="margin: 10px 0; color: #1a365d;">The LLM outputs a <strong>JSON object</strong>:</p>

<pre style="font-size: 14px; line-height: 1.5; margin: 10px 0; padding: 14px 18px; background: #ebf4ff; color: #1a365d; border-radius: 6px; border-left: 4px solid #3182ce; overflow-x: auto;"><code>{
  "name": "search_crop_disease",
  "args": {"disease": "Maize Smut", "crop": "maize", "language": "English"}
}</code></pre>

<p style="margin: 10px 0; color: #1a365d;"><strong>What you see:</strong> the tool call, the tool result, the final answer.</p>

<p style="margin: 10px 0; color: #1a365d;"><strong>What you don't see:</strong> <em>why</em> the LLM chose that tool.</p>

<h4 style="font-size: 16px; margin: 14px 0 6px 0; color: #2c5282;">Style 2 — ReAct agent (this notebook)</h4>

<p style="margin: 10px 0; color: #1a365d;">The LLM outputs <strong>structured prose</strong> in a strict format:</p>

<pre style="font-size: 14px; line-height: 1.5; margin: 10px 0; padding: 14px 18px; background: #ebf4ff; color: #1a365d; border-radius: 6px; border-left: 4px solid #3182ce; overflow-x: auto;"><code>Thought: The user asked about maize smut. I should search the knowledge base.
Action: search_crop_disease
Action Input: {'disease': 'Maize Smut', 'crop': 'maize', 'language': 'English'}
Observation: Plant resistant hybrids, rotate crops, remove galls before they burst.
Thought: I now have the treatment. I can answer.
Final Answer: To treat maize smut, plant resistant hybrids...</code></pre>

<p style="margin: 10px 0; color: #1a365d;"><strong>What you see:</strong> every thought, every action, every observation.</p>

<p style="margin: 10px 0; color: #1a365d;"><strong>Why it matters:</strong> you can <strong>audit</strong> the agent's reasoning. You see exactly why it did what it did.</p>

<h3 style="font-size: 19px; margin: 22px 0 10px 0; color: #2c5282;">The ReAct Loop in Detail</h3>

<p style="margin: 10px 0; color: #1a365d;">ReAct agents run this exact loop:</p>

<pre style="font-size: 14px; line-height: 1.5; margin: 10px 0; padding: 14px 18px; background: #ebf4ff; color: #1a365d; border-radius: 6px; border-left: 4px solid #3182ce; overflow-x: auto;"><code>1. Receive the user question
2. Output: Thought: [reasoning]
3. Output: Action: [tool name]
4. Output: Action Input: [arguments]
5. Your code runs the tool → gets Observation
6. Feed the Observation back to the LLM
7. Repeat 2-6 until the LLM outputs "Final Answer: ..."
8. Stop</code></pre>

<p style="margin: 10px 0; color: #1a365d;">The LLM keeps thinking until it has enough information. You see the whole journey.</p>
