```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
```

```python
# Load environment variables
load_dotenv()
```

```python
# Create model and prompt
llm = ChatOpenAI(model='gpt-4o-mini', temperature=0)
prompt = ChatPromptTemplate.from_messages([
    ('system', 'You are a helpful assistant.'),
    ('human', 'Explain {topic} in one sentence.')
])

# Build base chain
base_chain = prompt | llm | StrOutputParser()
```

Create the configured chain with metadata and tags, and invoke it

```python

# Using .with_config()
configured_chain = base_chain.with_config(
    run_name='RAG_Definition',
    metadata={'user_id': 'student_101', 'module': '2B'},
    tags=['advanced', 'lcel', 'with_config']
)

# Invoke the configured chain
result = configured_chain.invoke({'topic': 'RAG'})

print('Result')
print(result)
```
