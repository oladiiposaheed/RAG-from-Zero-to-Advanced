.bind() – Concept

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
```

```python
# Load environment variables
load_dotenv()

# Create a base chat model
base_llm = ChatOpenAI(model='gpt-4o-mini', temperature=0)

# Bind max_tokens=50 to create a short-answer model
short_llm = base_llm.bind(max_tokens=20)

# Bind max_tokens=200 to create a longer-answer model
long_llm = base_llm.bind(max_tokens=200)

# Create a prompt for both
prompt = ChatPromptTemplate.from_messages([
  ('system', 'You are a helpful assistant.'),
  ('human', 'Explain {topic} in two sentences.')  
])

# Build two chains using the bound models
chain_short = prompt | short_llm | StrOutputParser()
chain_long = prompt | long_llm | StrOutputParser()

# Invoke both chains with the same topic
result_short = chain_short.invoke({'topic': 'RAG'})
result_long = chain_long.invoke({'topic': 'RAG'})

print('Short model response:')
print(result_short)
print('\nLong model response:')
print(result_long)
```
