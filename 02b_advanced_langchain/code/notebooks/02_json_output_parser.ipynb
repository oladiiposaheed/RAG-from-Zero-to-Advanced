<div style="font-size: 0.85em;">
  <h3>JsonOutputParser – Straightforward Concept</h3>

  <p><strong>What is it?</strong><br>
  <code>JsonOutputParser</code> takes the model’s output (a string that contains JSON) and converts it into a Python dictionary.</p>

  <p><strong>Why use it?</strong></p>
  <ul>
    <li>Easily get structured data from the model without defining a full schema.</li>
    <li>Useful for simple key‑value outputs, like <code>{"answer": "...", "confidence": 0.95}</code>.</li>
    <li>Makes the model’s response ready for your application.</li>
  </ul>

  <p><strong>How it works:</strong></p>
  <ol>
    <li>You instruct the model to output JSON.</li>
    <li>The model returns a string, e.g. <code>'{"answer": "Paris", "confidence": 0.95}'</code>.</li>
    <li><code>JsonOutputParser</code> parses that string into a real Python dict:
      <code>{'answer': 'Paris', 'confidence': 0.95}</code>
    </li>
  </ol>

  <p><strong>Example flow:</strong></p>
  <pre style="font-family: monospace;">
Prompt → Model → JSON string → JsonOutputParser → Python dict
  </pre>
</div>

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import JsonOutputParser
from dotenv import load_dotenv
```

Step 2: Build a JSON-returning Chain

```python
# Create the chat model
llm = ChatOpenAI(model='gpt-4o-mini', temperature=0)

# Create a prompt that instructs the model to return JSON
prompt = ChatPromptTemplate.from_messages([
    ('system', 'You are a helpful assistant that always responds in valid JSON.'),
    ('human', 'Return a JSON object with keys "answer" and "confidence". The answer should be the capital of Nigeria, and confidence should greater than 0.90.')
])

# Build the chain: prompt -> model -> JsonOutputParser
chain = prompt | llm | JsonOutputParser()

# Invoke the chain
result = chain.invoke({})

print('Result:')
print(result)
print('Type:', type(result))
```
