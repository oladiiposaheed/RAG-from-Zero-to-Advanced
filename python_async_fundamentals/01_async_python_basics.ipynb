Running multiple coroutines concurrently using asyncio.gather. Show that total time is max, not sum.

```python
import time
import asyncio

async def do_task(name, seconds):
    print(f'{name} started')
    await asyncio.sleep(seconds)
    print(f'{name} finished after {seconds} seconds')
    
async def main():
    start = time.time()
    await asyncio.gather(
        do_task('A', 2),
        do_task('B', 3),
        do_task('C', 1)
    )
    end = time.time()
    print(f'Total time: {end - start:.2f} seconds')
    
await main()
```

Async with Lists, Loops, and If

```python
import time
import asyncio

# Define a list of task names and their durations
tasks_config = [
    ('Download', 2), 
    ('Process', 3), 
    ('Upload', 1), 
    ('Skip', 0)   # duration 0 means we will skip this one
]

# Define an async function that processes a task
async def run_task(name, seconds):
    if seconds <= 0:
        print(f'{name} skipped')
        return
    
    # Normal path: print start, sleep, print finish
    print(f'{name} started')
    await asyncio.sleep(seconds)
    print(f'{name} finished after {seconds} seconds')
    
# Define main coroutine
async def main():
    # Record start time
    start = time.time()
    
    # Create a list of coroutines by looping over tasks_config
    coroutines = []
    for name, sec in tasks_config:
        # Append each coroutine to the list
        coroutines.append(run_task(name, sec))
        
    # Run all coroutines concurrently
    await asyncio.gather(*coroutines)
    
    # Record end time
    end = time.time()
    print(f'Total time: {end - start:.2f} seconds')
    
await main()
```
