### Task 7: Error Handling and Robust Parsing

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import PydanticOutputParser
from langchain.output_parsers import OutputFixingParser
from pydantic import BaseModel, Field
from dotenv import load_dotenv
```

```python
# Load environment variables
load_dotenv()

# Define a simple Pydantic model
class PersonInfo(BaseModel):
    name: str = Field(description='Name of the person')
    age: int = Field(description='Age of the person')
    gender: str = Field(description='Male or female')
    
print('Imports and model defined.')
```

Create Parser and OutputFixingParser

```python
# Create the base Pydantic parser
parser = PydanticOutputParser(pydantic_object=PersonInfo)

# Create chat model
llm=ChatOpenAI(model='gpt-4o-mini', temperature=0)

# Wrap parser with auto-fix using LLM
fixing_parser = OutputFixingParser.from_llm(
    parser=parser,
    llm=llm
)
print('OutputFixingParser created.')
```

Simulate Bad Model Output and Parse It

```python
bad_output = 'name Fatimah age 25 gender female'

# Try normal parser
try:
    result = parser.parse(bad_output)
    print(f'Parsed normally: {result}')
    
except Exception as e:
    print('Normal parse error:')
    print(e)
    

# Try fixing parser
try:
    fixed_result = fixing_parser.parse(bad_output)
    print('\nFixed parse result:')
    print(fixed_result)
    print('Type:', type(fixed_result))
except Exception as e:
    print('Fixing parser also failed:')
    print(e)
```
