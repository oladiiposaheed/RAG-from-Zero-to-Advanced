### Task 3: Advanced Pydantic Features


 Define Nested and Advanced Models

```python
from pydantic import BaseModel, Field
from typing import Optional, List
```

```python
# Define an Author model
class Author(BaseModel):
    name: str
    email: str

# Define a Document model with nested Author
class Document(BaseModel):
    title: str = Field(description='Document title', min_length=1)
    author: Author = Field(description='Author of the document')
    tags: List[str] = Field(default=[], description='List of tags')
    pages: Optional[int] = Field(description='Number of pages', default=None)
```

Create a Valid Nested Instance

```python
# Create an Author
author = Author(name='Saheed', email='saheed@example.com')

# Create a Document with nested Author
doc = Document(
    title='RAG BootCamp',
    author=author,
    tags=['llm', 'rag', 'pydantic'],
    pages=850
)

print(doc)

print(f'Author type: {type(doc.author)}')
```

Optional Field Omitted

```python
# Create a document without pages
doc1 = Document(
    title='Introduction to RAG',
    author=author
)

print(doc1)
print(f'Pages: {doc1.pages}')
```
