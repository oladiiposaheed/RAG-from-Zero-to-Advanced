### Practical Example: RAG Document Metadata

**Why This Example?**

In a real RAG system, each retrieved document has metadata such as:

- Title
- Author
- Tags
- Language
- Number of pages
- Source URL

We use **Pydantic models** to validate and structure this metadata so that our pipeline doesn’t break when data is missing or malformed.

**What We’ll Do**

- Define `Author` and `KnowledgeArticle` models.
- Create valid articles with metadata.
- Simulate a retrieval list.
- Trigger a validation error.
- Display the retrieved data neatly.

```python
from typing import List, Optional
from pydantic import BaseModel, Field
```

Model for an article author

```python
class Author(BaseModel):
    name: str
    email: str
    
# Model for a knowledge article
class KnowledgeArticle(BaseModel):
    title: str = Field(description='Title of the article', min_length=2)
    author: Author = Field(description='Who wrote the article')
    tags: List[str] = Field(default=[], description='List of topics/tags')
    language: str = Field(default='English', description='Language of the article')
    pages: Optional[int] = Field(default=None, description='Number of pages if PDF')
    url: Optional[str] = Field(default=None, description='Original dource URL')
    
print('Models created successfully.')
```

Create Valid Articles


```python
# Create authors
author1 = Author(name='Saheed', email='saheed@example.com')
author2 = Author(name='Pato', email='pato@example.com')

# Create articles
article1 = KnowledgeArticle(
    title='How to Use RAG for Customer Support',
    author=author1,
    tags=['llm', 'rag', 'agentic-ai', 'customer-support'],
    language='English',
    pages=230,
    url='https://internal.company.com/docs/rag-support'
)

article2 = KnowledgeArticle(
    title='Guide to Vector Databases',
    author=author2,
    tags=['vector-db', 'faiss', 'pinecone'],
    language='English'
)

print('Article 1:')
print(article1)
print('\nArticle 2:')
print(article2)
```

Simulate Retrieval and Display

```python
# Simulate a list of retrieved documents
retrieved_docs = [article1, article2]

# Loop and display metadata 
for doc in retrieved_docs:
    print(f'Title: {doc.title}')
    print(f'Author: {doc.author.name} ({doc.author.email})')
    print(f'Tags: {", ".join(doc.tags)}')
    print(f'Language: {doc.language}')
    
    if doc.pages:
        print(f'Pages: {doc.pages}')
        
    if doc.url:
        print(f'Source: {doc.url}')
        print('-' * 60)
```

Validation Error Example


```python
# Invalid article 
try:
    invalid_Article = KnowledgeArticle(
        title='',
        author=author1,
        tags=['test']
    )
    
except Exception as e:
    print(f'Error type: {type(e).__name__}')
    print(e)
```
