### Task 1: Pydantic Models and Validation

```python
# Import and Define a Model

from pydantic import BaseModel

# Define a simple Pydantic model
class Person(BaseModel):
    name: str
    age: int
```

```python
# Create a Valid Instance
person1 = Person(name='Fatimah', age=23)

print(f'Person 1: {person1}')
print(f'Type of name: {type(person1.name)}')
print(f'Type of age: {type(person1.age)}')
```

Automatic Type Conversion

```python
# Create Person with age as a string
person2 = Person(name='Peter', age='15')
print(f'Person2: {person2}')
print(f'Type of person2.age: {type(person2.age)}')
```

Trigger a Validation Error

```python
try:
    person3 = Person(name='Chika', age='thirty two')

except Exception as e:
    print(f'Error: {type(e).__name__}')
    print(e)
```
