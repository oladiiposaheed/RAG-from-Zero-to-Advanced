### Task 2: Field Descriptions and Defaults

```python
from pydantic import BaseModel, Field

class Product(BaseModel):
    name: str = Field(description='Name of the product', min_length=2)
    price: float = Field(description='Price in Naira', gt=0)
    category: str = Field(description='Product category', default='general')
    
print('Product model defined.')
```

Create a Valid Instance

```python
product = Product(name='Milk', price='2000.50')

print(product)
print(f'Category: {product.category}')
```

Violate a Constraint

```python
try:
    bad_product = Product(name='Bread', price=0)

except Exception as e:
    print(f'Error type: {type(e).__name__}')
    print(e)
```
