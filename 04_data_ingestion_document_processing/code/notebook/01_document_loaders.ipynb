<div style="font-size: 0.85em;">
  <h3>Document Loaders</h3>
  <p>A <strong>Document Loader</strong> is a component that reads data from a source (PDF, CSV, TXT, JSON, HTML, etc.) and converts it into LangChain <code>Document</code> objects.</p>
  <p>Each <code>Document</code> has:</p>
  <ul>
    <li><code>page_content</code> – the actual text</li>
    <li><code>metadata</code> – extra info like source, file name, page number</li>
  </ul>
  <p>Why use loaders?</p>
  <ul>
    <li>Standardise different file formats</li>
    <li>Preserve metadata for later retrieval</li>
    <li>Prepare documents for splitting and embedding</li>
  </ul>
  <p>In this notebook, we will load:</p>
  <ul>
    <li>PDF files (<code>PyPDFLoader</code>)</li>
    <li>Text files (<code>TextLoader</code>)</li>
    <li>CSV files (<code>CSVLoader</code>)</li>
    <li>JSON files (<code>JSONLoader</code>)</li>
    <li>HTML files (<code>UnstructuredHTMLLoader</code> or <code>BSHTMLLoader</code>)</li>
  </ul>
</div>

Step 6b: Load JSON with JSONLoader

```python
# Path to JSON file
json_path = '../../data/ecommerce.json'

# Define a function to extract metadata
def metadata_func(record: dict, metadata: dict) -> dict:
    
    metadata['userId'] = record.get('userId')
    metadata['id'] = record.get('id')
    metadata['title'] = record.get('title')
    
    return metadata

# Create JSONLoader with jq
loader = JSONLoader(
    file_path=json_path,
    jq_schema='.[]',
    content_key='body',
    metadata_func=metadata_func
)

# Load documents
docs_json = loader.load()

print('First document metadata:')
print(f'{docs_json[0].metadata}')
```

```python
import re

# Path to HTML file
html_path = '../../data/agriculture.html'

# Create loader
loader_html = BSHTMLLoader(
    file_path=html_path,
    open_encoding='utf-8',
    bs_kwargs={'features': 'html.parser'}
)

# Load document
docs_html = loader_html.load()

# Clean the page content
raw_text = docs_html[0].page_content

# Replace multiple newlines with a single newline
clean_text = re.sub(r'\n+', '\n', raw_text)

# Remove excessive spaces at the beginning of lines
clean_text = re.sub(r'[\t]+', ' ', clean_text)

# Strip loading/trailing whitespace
clean_text = clean_text.strip()

print('Cleaned content preview:')
print(clean_text[:1500])

print('\nMetadata:')
print(docs_html[0].metadata)
```

```python
def load_files(file_path):
    ext = os.path.splitext(file_path)[1].lower()
    if ext == '.txt':
        return TextLoader(file_path, encoding='utf-8').load()
    elif ext == '.csv':
        return CSVLoader(file_path, encoding='utf-8').load()
    elif ext == '.json':
        # Do NOT pass encoding here
        return JSONLoader(
            file_path=file_path,
            jq_schema='.[]',
            content_key='body'
        ).load()
    elif ext == '.pdf':
        return PyPDFLoader(file_path).load()
    elif ext == '.html':
        return BSHTMLLoader(
            file_path=file_path,
            open_encoding='utf-8',
            bs_kwargs={'features': 'html.parser'}
        ).load()
    else:
        return []
```

```python
all_docs = []

for f in os.listdir(data_dir):
    path = os.path.join(data_dir, f)
    if os.path.isfile(path):
        docs = load_files(path)
        all_docs.extend(docs)

# Display documents
for i, doc in enumerate(all_docs, start=1):
    print(f'Document {i}:')
    print(f"Source: {doc.metadata.get('source', 'unknown')}")
    print(f"Title: {doc.metadata.get('title', 'N/A')}")
    print(f"Page: {doc.metadata.get('page', 'N/A')}")
    print(f"Row: {doc.metadata.get('row', 'N/A')}")
    print(f"Seq: {doc.metadata.get('seq_num', 'N/A')}")
    print('-' * 60)
    lines = doc.page_content.split('\n')
    clean_lines = [line.strip() for line in lines if line.strip()]
    clean_text = '\n'.join(clean_lines)
    print(clean_text[:500])
    print('=' * 100)
    print()
```
