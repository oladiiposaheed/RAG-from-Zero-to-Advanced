### Task 5: Chunk Size, Overlap, and Metadata

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document
```

Create a sample document

```python
sample_text = '''
Agriculture is very important in Nigeria. Agriculture provides food for millions.
Agriculture employs many people in rural areas. Agriculture contributes to GDP.
'''

sample_doc = Document(page_content=sample_text)

print(f'Sample text length: {len(sample_doc.page_content)}')
```

Create different chunk sizes with no overlap

```python
for size in [20, 50, 100]:
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=size,
        chunk_overlap=0,
        separators=['\n\n', '\n', '.', ' ', '']
    )
    chunks = splitter.split_documents([sample_doc])
    print(f'chunk_size={size} -> {len(chunks)} chunk(s)')
```

Test Different Overlap Values with fixed chunk size

```python
for overlap in [0, 10, 25]:
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=50,
        chunk_overlap=overlap,
        separators=['\n\n', '\n', '.', ' ', '']
    )
    
    chunks = splitter.split_documents([sample_doc])
    print(f'chunk_overlap={overlap} -> {len(chunks)} chunk(s)')
```

View Chunk Contents with Overlap

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=50,
    chunk_overlap=25,
    separators=['\n\n', '\n', '.', ' ', '']
)

chunks = splitter.split_documents([sample_doc])

for i, chunk in enumerate(chunks, start=1):
    print(f'Chunk {i}:')
    print(chunk.page_content)
    print('-' * 50)
```

Create a Longer Sample Text

```python
# Longer sample text about agriculture
long_text = """
Agriculture is one of the most important sectors in Nigeria. It provides food for millions of people and raw materials for industries. Many families depend on farming for their daily income. The main crops grown include cassava, yam, maize, rice, sorghum, millet, and beans.

Cassava is processed into garri, fufu, and starch. Yam is a major staple in the middle belt and southern regions. Maize is used for both human consumption and animal feed. Rice production has increased due to government policies. Sorghum and millet are common in the north where rainfall is low.

Livestock farming is also widespread. Cattle, goats, sheep, and poultry are raised in many communities. Fisheries and aquaculture are growing rapidly and providing protein and employment.

However, agriculture faces challenges such as poor roads, limited credit, climate change, and post-harvest losses. Despite these problems, agriculture remains vital for food security and economic growth.
"""

long_doc = Document(page_content=long_text)
print('Long text length:', len(long_doc.page_content))
```

Test Overlap with Longer Text

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=50,
    separators=['\n\n', '\n', '.', ' ', '']
)

chunks = splitter.split_documents([long_doc])

for i, chunk in enumerate(chunks[:4], start=1):
    print(f'Chunk {i}')
    print(chunk.page_content)
    print('-' * 60)
```

```python
# Use a simple CharacterTextSplitter with separator=''
from langchain_text_splitters import CharacterTextSplitter

splitter = CharacterTextSplitter(
    separator='',         # split by pure characters
    chunk_size=100,
    chunk_overlap=30      # 30 characters shared
)

chunks = splitter.split_documents([long_doc])

print(f'Number of chunks: {len(chunks)}\n')

# Print first 3 chunks and highlight the overlap
for i, chunk in enumerate(chunks[:3], start=1):
    print(f'Chunk {i} (length={len(chunk.page_content)}):')
    print(chunk.page_content)
    print('-' * 60)
```
