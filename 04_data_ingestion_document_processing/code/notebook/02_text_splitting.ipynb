### Text Splitting Strategies

Step 1: Imports

````python
from langchain_core.documents import Document
from langchain_text_splitters import (
    CharacterTextSplitter,
    RecursiveCharacterTextSplitter
)
from langchain_community.document_loaders import TextLoader
````

Step 2: Create a Sample Document

````python
sample_text = """
Agriculture is the backbone of many economies around the world, 
especially in Africa. In Nigeria, agriculture employs a large portion of the
population and contributes significantly to the country's Gross Domestic Product.
The main crops grown in Nigeria include cassava, yam, maize, rice, sorghum, millet, 
and beans. Cassava is one of the most important staple crops, processed into garri,
fufu, and starch. Yam is another major tuber crop, particularly in the middle belt and
southern regions. Maize is widely cultivated for both human consumption and animal feed.
Rice production has increased in recent years due to government policies aimed at 
reducing imports. Sorghum and millet are common in the northern states where rainfall
is lower. Beans provide an important source of protein for many households. 
Livestock farming, including cattle, goats, sheep, and poultry, is also widespread.
In addition, fisheries and aquaculture have grown rapidly, providing employment 
and protein. However, agriculture in Nigeria faces several challenges, including poor
infrastructure, limited access to credit, climate change, and post-harvest losses. 
Despite these challenges, the sector remains vital for food security and economic growth.
"""
````

````python
# Create a LangChain Document object from the sample text
sample_doc = Document(page_content=sample_text)

sample_doc
````

````python
# Print the length of the document content
print(f'Sample document length: {len(sample_doc.page_content)} characters')
print('\nPreview')
print(sample_doc.page_content[:100])
````

Step 3: Split with CharacterTextSplitter

````python
char_splitter = CharacterTextSplitter(
    chunk_size=200, chunk_overlap=50, separator=''
)

# Split the sample document into chunks
char_chunks = char_splitter.split_documents([sample_doc])

# Number of chunk
print(f'Number of chunks: {len(char_chunks)}')

# Extract each chunk with its length
for i, chunk in enumerate(char_chunks, start=1):
    print(f'\n--- Chunk {i} (length={len(chunk.page_content)}) ---')
    print(chunk.page_content)
````

Step 4: Split with RecursiveCharacterTextSplitter


````python
recursive_splitter = RecursiveCharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=50
)

# Split the sample document
recursive_chunks = recursive_splitter.split_documents([sample_doc])

# Print number of chunks
print(f'Number of chunks: {len(recursive_chunks)}')

# Print each chunk with length
for i, chunk in enumerate(recursive_chunks, start=1):
    print(f'\n--- Chunk {i} (length={len(chunk.page_content)}) ---')
    print(chunk.page_content)
````

Step 5: Split a Real Document

````python
# Path to the text file
txt_path = '../../data/agriculture.txt'

# Load the document
loader = TextLoader(txt_path, encoding='utf-8')
docs = loader.load()


print(f'Loaded {len(docs)} document(s)')

# Split the loaded document

recursive_splitter = RecursiveCharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=50,
    separators=['\n\n', '\n', '.', ' ', '']
)
split_docs = recursive_splitter.split_documents(docs)

# Print number of chunks
print(f'Number of chunks after splitting: {len(split_docs)}')

# Print first 3 chunks
for i, chunk in enumerate(split_docs[:3], start=1):
    print(f'\n--- Chunk {i} (length={len(chunk.page_content)}) ---')
    print(chunk.page_content)
````

Step 6: Use the Improved Splitter for Real Documents


````python
r_splitter = RecursiveCharacterTextSplitter(
    chunk_size=300,
    chunk_overlap=50,
    separators=['\n\n', '\n', '.', ' ', '']
)

# Split the same document
r_chunks = r_splitter.split_documents(docs)

print(f'Number of chunks: {len(r_chunks)}')

# Show first 5 chunks with lengths
for i, chunk in enumerate(r_chunks[:5], start=1):
    print(f'\n--- Chunk {i} (length={len(chunk.page_content)}) ---')
    print(chunk.page_content)
````

Step 7: Preserve Metadata During Splitting

````python
# Show metadata of the first few chunks
for i, chunk in enumerate(r_chunks[:3], start=1):
    print(f'Chunk {i} metadata:')
    print(chunk.metadata)
    print('-' * 70)
````

Step 8: Add Custom Metadata to Chunks

````python
for i, chunk in enumerate(r_chunks):
    chunk.metadata['doc_type'] = 'text'
    chunk.metadata['language'] = 'English'
    chunk.metadata['chunk_id'] = f'agriculture_txt_{i+1:03d}'
    
# Show updated metadata for first 3 chunks
for i, chunk in enumerate(r_chunks[:3], start=1):
    print(f'Chunk {i} metadata:')
    print(chunk.metadata)
    print('-' * 50)
````

Step 9: Save Chunks for Later Use

````python
import json
import os

# Convert chunks to a list of dict
chunk_data = []

for chunk in r_chunks:
    chunk_data.append({
        'content': chunk.page_content,
        'metadata': chunk.metadata
    })
    
# Save to JSON
output_path = '../../data/chunks/agriculture_chunks.json'

# Create directory if not exists
os.makedirs(os.path.dirname(output_path), exist_ok=True)

# Open the file in write mode
with open(output_path, 'w', encoding='utf-8') as f:
    json.dump(chunk_data, f, indent=2, ensure_ascii=False)
    
print(f'Save {len(chunk_data)} chunks to {output_path}')
````

Step 10: Load the Saved Chunks



````python
# Path to the saved chunks file
chunks_path = '../../data/chunks/agriculture_chunks.json'

# Open and load the JSON file
with open(chunks_path, 'r', encoding='utf-8') as f:
    loaded_chunks = json.load(f)
    
# Print total number of chunks
print(f'Loaded {len(loaded_chunks)} chunks from JSON')

# Print the first chunk 
print('\nFirst chunk content:')
print(loaded_chunks[0]['content'])

print('\nFirst chunk metadata:')
print(loaded_chunks[0]['metadata'])
````

## Summary: Text Splitting Strategies

### What We Learned

- **Why split documents?**
  - LLMs have token limits.
  - Smaller chunks improve retrieval accuracy.
  - Easier to store and search in vector databases.

- **CharacterTextSplitter**
  - Splits by raw character count.
  - Can be forced with `separator=''`.
  - Often too crude for natural text.

- **RecursiveCharacterTextSplitter** (most common)
  - Splits by natural boundaries: paragraphs, lines, sentences, words.
  - Preserves readability and context.
  - Best for general RAG use.

- **Chunk Size and Overlap**
  - `chunk_size` = maximum size of each chunk.
  - `chunk_overlap` = amount shared between adjacent chunks.
  - Overlap prevents context loss at boundaries.

- **Metadata**
  - Each chunk inherits metadata from its parent document.
  - We added custom metadata: `doc_type`, `language`, `chunk_id`.
  - Metadata is essential for filtering and citation later.

- **Saving Chunks**
  - We saved chunks to JSON for reuse in later modules.
  - JSON format stores both `content` and `metadata`.
